```python
total_recruitment_costs = recruitment_costs.groupby('Employment Source')['Total'].sum()

performance_by_source = staff_particulars.groupby('Employee Source')['Performance Score'].value_counts().unstack().fillna(0)

decent_performance_labels = ['Fully Meets', 'Exceeds', 'Exceptional', '90-day meets']
performance_by_source['Decent Performance'] = performance_by_source[decent_performance_labels].sum(axis=1)
performance_by_source['All Performance'] = performance_by_source[all_performance_labels].sum(axis=1)

combined_data = pd.DataFrame({
    'Total Cost': total_recruitment_costs,
    'Decent Performance Count': performance_by_source['Decent Performance'],
    'All Performance Count': performance_by_source['All Performance']
}).dropna()

# Calculate a simple metric for cost-effectiveness (Decent Performance Count / Total Cost)
combined_data['Proportion of Decent Performers'] = combined_data['Decent Performance Count'] / combined_data['All Performance Count'] * 100

combined_data['Cost Effectiveness - Decent'] = combined_data['Decent Performance Count'] / combined_data['Total Cost'] * 1000
combined_data['Cost Effectiveness - All'] = combined_data['All Performance Count'] / combined_data['Total Cost'] * 1000

sorted_combined_data = combined_data.sort_values(by='All Performance Count', ascending=False)

sorted_combined_data[sorted_combined_data['Total Cost'] == 0].head(5)
```

```python
decent_performance_labels = ['Fully Meets', 'Exceeds', 'Exceptional', '90-day meets']
all_performance_labels = ['Fully Meets', 'Exceeds', 'Exceptional', '90-day meets', "PIP", "Needs Improvement"]

total_recruitment_costs = recruitment_costs_data.groupby('Employment Source')['Total'].sum()

performance_by_source = staff_performance_data.groupby('Employee Source')['Performance Score'].value_counts().unstack().fillna(0)

performance_by_source['All Performance'] = performance_by_source[all_performance_labels].sum(axis=1)
performance_by_source['Decent Performance'] = performance_by_source[decent_performance_labels].sum(axis=1)

# Create a DataFrame for comparison
combined_comparison = pd.DataFrame({
    'Total Cost': total_recruitment_costs,
    'Decent Performance Count': performance_by_source['Decent Performance'],
    'All Performance Count': performance_by_source['All Performance']
}).fillna(0)

# Calculate Proportion of Decent Performers
combined_comparison['Proportion of Decent Performers'] = combined_comparison['Decent Performance Count'] / combined_comparison['All Performance Count'] * 100

combined_comparison['Cost Effectiveness - Decent'] = combined_comparison['Decent Performance Count'] / combined_comparison['Total Cost'] * 1000
combined_comparison['Cost Effectiveness - All'] = combined_comparison['All Performance Count'] / combined_comparison['Total Cost'] * 1000

sorted_comparison = combined_comparison.sort_values(by='Cost Effectiveness - Decent', ascending=False)

# Proportion of Decent Performers
filted_comparison = sorted_comparison[sorted_comparison['Decent Performance Count'] > 5]
filtered_comparison = filted_comparison[filted_comparison['Total Cost'] > 0]

filtered_comparison.head(5)
```

<p style="

">
    Employees whose performance score have not been reviewed are also included
</p>


```python
performance_scores = {
    'PIP': 0,
    '90-day meets': 1,
    'Needs Improvement': 0,
    'Fully Meets': 2,
    'Exceeds': 3,
    'Exceptional': 4
}

staff_particulars['Performance Score Numeric'] = staff_particulars['Performance Score'].map(performance_scores)

average_performance_by_manager = staff_particulars.groupby('Manager Name')['Performance Score Numeric'].mean()

total_performance_by_manager = staff_particulars.groupby('Manager Name')['Performance Score Numeric'].sum()
max_score_per_employee = max(performance_scores.values())
total_possible_performance_by_manager = staff_particulars.groupby('Manager Name').size() * max_score_per_employee

# Get the manager's own performance score
managers_own_performance = staff_particulars.drop_duplicates('Manager Name').set_index('Manager Name')['Performance Score Numeric']

staff_particulars['Terminated'] = staff_particulars['Date of Termination'].notna()
retention_rate_by_manager = staff_particulars.groupby('Manager Name')['Terminated'].apply(lambda x: (1 - x.mean()) * 100)

manager_metrics = pd.DataFrame({
    'Average Performance Score': average_performance_by_manager,
    'Manager Own Performance Score': managers_own_performance,
    'Total Team Performance Score': total_performance_by_manager,
    'Total Possible Team Performance Score': total_possible_performance_by_manager,
    'Retention Rate': retention_rate_by_manager,
})

def manager_score(row):
    team_perf_score = (row['Total Team Performance Score'] / row['Total Possible Team Performance Score']) * 0.40
    manager_perf_score_ratio = (row['Manager Own Performance Score'] / 4) * 0.20
    retention_score = (row['Retention Rate'] / 100) * 0.40
    score = team_perf_score + manager_perf_score_ratio + retention_score
    return score


manager_metrics['Overall Score'] = manager_metrics.apply(manager_score, axis=1)

best_manager = manager_metrics['Overall Score'].idxmax()
worst_manager = manager_metrics['Overall Score'].idxmin()

# best_manager, manager_metrics.loc[best_manager], worst_manager, manager_metrics.loc[worst_manager]
manager_metrics['Overall Score']
```

```python
reasons_for_termination = staff_particulars['Reason For Term'].dropna()

grouped_reasons = {
    'Personal/Career Development': ['Another position', 'more money', 'career change', 'military'],
    'Job Dissatisfaction': ['unhappy', 'hours'],
    'School' : ['return to school'],
    'Performance related' : ['gross misconduct', 'no-call, no-show', 'attendance', 'performance'],
    'Other Reasons': ['medical issues', 'maternity leave - did not return', 'relocation out of area', 'retiring']
}


all_grouped_reasons = sum(grouped_reasons.values(), [])

filtered_reasons_for_termination = reasons_for_termination[reasons_for_termination.isin(all_grouped_reasons)]

filtered_reasons_mapped = filtered_reasons_for_termination.apply(lambda x: next((key for key, values in grouped_reasons.items() if x in values), 'Other Reasons'))

filtered_grouped_reason_counts = filtered_reasons_mapped.value_counts()

plt.figure(figsize=(8, 8))
plt.pie(filtered_grouped_reason_counts, labels=filtered_grouped_reason_counts.index, autopct='%1.1f%%')
plt.title('Reasons for Termination (Grouped)')
plt.show()
```
